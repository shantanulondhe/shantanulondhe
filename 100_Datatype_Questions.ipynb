```python

def is_anagram(str1, str2):
    clean1 = "".join(str1.lower().split())
    clean2 = "".join(str2.lower().split())
    return sorted(clean1) == sorted(clean2)
```

```python
print(is_anagram("Listen", "Silent"))          # True
```

```python
#18. Simple email validator, Use basic string operations and conditions to check for a plausible email structure: one @, a nonempty local part, and a domain containing a dot. This is a basic exercise, not full email-standard validation.
email=input('enter email: ')
atcount=email.count('@')
if atcount!=1:
  print('Atleast 1 @')
else:
  localname, domainname=email.split('@')
  if len(localname)==0:
    print('No local name')
  elif '.' not in domainname:
    print('Domain name dosent contain dot(.)')
  elif domainname.endswith('.') or domainname.startswith('.')==True:
    print('Email cnannot starts with or ends with (.)')
  else:
    print('Email Created successfully')
```

```python
'''20. Mini NLP challenge — Most informative words, Given a document, calculate word frequencies and
return the three most frequent words. Ignore common stop words such as "the", "is", and "and". Resolve ties consistently.'''

para='''Did you finish the quarterly report?" asked Clara, rubbing her tired eyes; "we need—now more than ever—to meet
        today's tight deadline!" He replied quickly, "Yes, all the charts, figures, and summaries are ready; however,
        do we have the manager's final approval?" With a deep sigh, she checked her email inbox, whispered "Let's hope so," and
        clicked send.'''

count=0
dupwd=[]
uwd=[]
para=para.split()
for i in para:
  if i in uwd:
    if i not in dupwd:
      count+=1
      dupwd.append(i)
  else:
    uwd.append(i)

print(dupwd)
print(count)
```

```python
def printf(*args, sep=" "):
    # Join all items together as strings, separated by the 'sep' character
    output = sep.join(str(item) for item in args)

    # Send the final string to the built-in print
    print(output)
```
