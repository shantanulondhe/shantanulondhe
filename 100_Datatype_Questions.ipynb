```python
#1. Count vowels and consonants Given a string, count vowels and consonants separately. Ignore spaces, numbers, and punctuation.
str=input('Enter a string: ')
vowels=['a','e','i','o','u']

unwant=[1,2,3,4,5,6,7,8,9,0,'@','#','$','%','+','+',' ']
count=0
count1=0
for i in str.lower():
  if i.isalpha():
    if i in vowels:
      count+=1
      #print(i,' is vowel')
    else:
      count1+=1
      #print(i,' is consonent')
print(count)
print(count1)
```

```python
# 2. Character frequency Count how many times each character appears, ignoring spaces and case.
str=input('Enter a string: ')
samechar={}
count=0
for i in str.lower():
  if i.isalnum():
    if i in samechar:
      samechar[i] += 1
    else:
      samechar[i] = 1
print("Character frequencies:")
print(samechar)
```
