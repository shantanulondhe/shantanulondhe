```python
#31. Check whether a number is prime.
num=79
if num > 1:
    for i in range(2, int(num**0.5) + 1):
      if num % i == 0:
         print(f"{num} is NOT a prime number.")
         break
    else:
        print(f"{num} is a prime number.")
else:
    print(f"{num} is NOT a prime number.")
```

```python
#33 Print all prime numbers between 1 and 100.
primes = [num for num in range(1, 101) if num > 1 and all(num % i != 0 for i in range(2, int(num**0.5) + 1))]

print(primes)
```
