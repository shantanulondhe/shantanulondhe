```python
#23. Find common elements in two lists
def comm_eliminate(l1,l2):
  comm=[]
  for i in l1:
    for j in l2:
      if i==j:
        comm.append(i)
  return comm
comm_eliminate([1,2,3,4,9],[3,2,9,1,6])
```

```python
#25 25. Return all prime numbers up to n
def all_prime(num):
  prime=[]
  for i in range (2,num+1):
    if is_prime(i):
      prime.append(i)

  return prime
all_prime(11)
```

```python
#32. Accept any number of positional arguments using *args
def add(*add):
  total=0
  for i in add:
    total+=i
  return total
add(2,3,4,5,6)
```

```python
#39. Fibonacci sequence using a function
def fibonacci(n):
    result = []
    a, b = 0, 1

    for _ in range(n):
        result.append(a)
        a, b = b, a + b

    return result

print(fibonacci(10))
```

```python
#40. Validate a password
def validate_pass(pas):
  if len(pas)<8:
    return False
  has_upper = any(ch.isupper() for ch in pas)
  has_lower=any (ch.islower()for ch in pas)
  has_digit=any (ch.isdigit()for ch in pas)

  return has_upper and has_lower and has_digit

validate_pass('shantanuS9')
```
