# map(), lambda, filter(), reduce(), and higher-order functions.::

```python
# Example 1 — Double each number
num=[1,2,3,4,5]
result= list(map(lambda x: x*2, num ))

print(result)
```

```python
#Example 2 — Square each number
num=[1,2,3,4,5]
result= list(map(lambda x: x**2, num))

print(result)
```

```python
#Example 3 — Convert strings to uppercase
str='shantanu'
result= list(map(lambda x: x.upper(),str))
result= ''.join(result)
print(result)
```

```python
#Example 4 — Convert strings to integers
l=['1','2','3','4']
result=list(map(int,l))
print(result)
```

```python
#Example 5 — Add 10 to each number
l=[1, 2, 3, 4]
result= list(map(lambda x: x+10, l))
print(result)
```

```python
#Example 7 — Add corresponding elements from two lists
l=[1, 2, 3, 4]
l1=[10,20,30,40]
result= list(map(lambda x,y: x+y, l,l1))
print(result)
```

```python
#Example 8 — Calculate final prices after 18% tax
l1=[10,20,30,40]
result = list(map(lambda x: x + x*0.18 ,l1))
print(result)
```

```python
#Example 9 — Extract names from dictionaries
students = [
    {"name": "Amit", "marks": 80},
    {"name": "Riya", "marks": 90},
    {"name": "Sam", "marks": 75}
]
result=', '.join(map(lambda x: x['name'], students))
print(result)
```

```python
#Example 10- * diagrams
draw_pyramid = lambda rows: print("\n".join(" " * (rows - i) + "* " * i for i in range(1, rows + 1)))

draw_pyramid(5)
```

2. lambda — 10 examples

```python
#Example 1 — Add multiple numbers

result = lambda *x: sum(x)
print(result(20,2,7,98,90))
```

```python
#Example 2 — Multiply two numbers
multiply = lambda a, b: a * b
print(multiply(5, 6))
```

```python
#Example 3 — Check even or odd
even_odd= lambda x :  'even' if x%2 ==0  else 'odd'
print(even_odd(9))
```

```python
#Example 4 — Find the larger number
l1=[20,54,802,279,999,3219]
largest= list(map(lambda x: x if x>x else l1, l1))
print(largest())
```
