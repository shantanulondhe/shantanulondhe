```python
# List: find max/min/sum/average without built-ins
l1=[22,86,90,33,54,2,1]
total=0
maxs=max(l1)
print('Max using builtin function: ',maxs)

l1.sort()
print("The maximum value using sorting method is: ", l1[-1],'\n')

mins=min(l1)
print('Min using builtin function: ',mins)
print("The minimum value using sorting method is: ", l1[0],'\n' )

#sum
for i in l1:
  total+= i
print('Sum of all numbers in list is: ',total)
print(sum(l1))

#average

avg=sum(l1)/len(l1)
print ('Using built in function: ',avg)

#avg using loop
totsum=0
count=0
for num in l1:
  totsum+=num
  count+= 1
if count>0:
  average = totsum / count
else:
    average = 0

print("The average is:", average)


total = 0
for num in l1: total += num

average = total / len(l1)
print("Average:", average)
```
